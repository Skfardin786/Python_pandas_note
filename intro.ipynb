# Reindex

```python
sales_jan =pd.DataFrame({
    'Sales': [250, 400, 300]   
}, index=['Store_A','Store_B','Store_C'])

sales_jan
```

```python
sales_feb =pd.DataFrame({
    'Sales': [260, 320]   
}, index=['Store_A','Store_C'])

sales_feb
```

```python
common_index = sales_jan.index
```

```python
sales_feb_aligned = sales_feb.reindex(common_index)
```

```python
sales_feb_aligned.head()
```

```python
# fill value in NaN position.
sales_feb_aligned = sales_feb.reindex(common_index, fill_value=0)
```

```python
sales_feb_aligned
```

```python
sales_diff = sales_feb_aligned['Sales'] - sales_jan['Sales']

sales_diff
```

# Condition based Column creation using custom logic

```python
df1.head()
```

```python
# if >= 30, senior else young

values = []

for idx, row in df1.iterrows():
    if row['Age'] < 30:
        values.append('young')
    else:
        values.append('senior')

df1['Category'] = values
```

```python
df1
```

```python
# again Print the dataframe.
df2 = pd.DataFrame({
    'Name' : ['fardin','istyak','nadim'],
    'Age' : [22,26,29],
    'City' : ['Jajpur','Baruan','Brahmabarada']
})
```

```python
# vectorized using apply

df2['Categories'] = df2['Age'].apply(lambda age : 'young' if age < 30 else 'senior')
```
